# SQL – práce s databází store


## 1. Jaká data jsou ve sloupci description a jakou mají strukturu?

Sloupec `description` je datového typu **JSON**.

Obsahuje strukturovaná data popisující vlastnosti produktu.

JSON je tvořen dvojicemi:

klíč : hodnota

Hodnoty mohou být:
- text (string)
- čísla
- boolean hodnoty

Typicky obsahuje klíč `category`, který určuje typ produktu.

```python
-- Ukázka obsahu JSON sloupce
SELECT description
FROM product;
```

## 2. Co provede následující dotaz

Dotaz vybere název produktu a z JSON sloupce `description` načte hodnotu `category`.

Funkce `JSON_VALUE` umožňuje číst hodnoty z JSON objektu.

`$.category` znamená cestu k položce `category` v JSON objektu.

```python
SELECT
name,
JSON_VALUE(description, '$.category') AS category
FROM product;
```

## 3. SQL pro zobrazení všech produktů z vybraného skladu

Například zobrazíme produkty ze skladu s ID = 1.

```python
SELECT *
FROM product
WHERE warehouse_id = 1;
```

## 4. SQL pro zobrazení všech produktů s kategorií cable

```python
SELECT *
FROM product
WHERE JSON_VALUE(description, '$.category') = 'cable';
```

## 5. SQL pro zobrazení produktů s počtem větším než 20 seřazených vzestupně

```python
SELECT *
FROM product
WHERE quantity > 20
ORDER BY quantity ASC;
```

## 6. SQL pro produkty uskladněné v březnu 2024

```python
SELECT *
FROM product
WHERE stored_from BETWEEN '2024-03-01' AND '2024-03-31';
```
